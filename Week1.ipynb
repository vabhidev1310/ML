```python
# Program 1: Variables and Data Types

# 1. Declare variables of different data types
num = 25                # Integer
price = 99.99           # Float
name = "Snikitha"       # String
is_student = True       # Boolean

# Display values and their data types
print("Initial Variables:")
print("num =", num, "| Data Type:", type(num))
print("price =", price, "| Data Type:", type(price))
print("name =", name, "| Data Type:", type(name))
print("is_student =", is_student, "| Data Type:", type(is_student))

print("\nChecking Data Types using type():")
print(type(num))
print(type(price))
print(type(name))
print(type(is_student))

# 2. Modify the values of the variables
num = 25.5
price = "One Hundred"
name = False
is_student = 1

# Display modified values and data types
print("\nAfter Modifying Variables:")
print("num =", num, "| Data Type:", type(num))
print("price =", price, "| Data Type:", type(price))
print("name =", name, "| Data Type:", type(name))
print("is_student =", is_student, "| Data Type:", type(is_student))

# 3. Student details
student_name = "abhi"
student_age = 20
student_marks = 88.5
student_pass = True

# Display student details and their data types
print("\nStudent Details:")
print("Name:", student_name, "| Data Type:", type(student_name))
print("Age:", student_age, "| Data Type:", type(student_age))
print("Marks:", student_marks, "| Data Type:", type(student_marks))
print("Passed:", student_pass, "| Data Type:", type(student_pass))
```

```python
# Program 2: Operators

# 1. Arithmetic Operators
a = 20
b = 5

print("Arithmetic Operators")
print("a =", a, "b =", b)
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)

# 2. Comparison Operators
print("\nComparison Operators")
print("a == b :", a == b)
print("a != b :", a != b)
print("a > b :", a > b)
print("a < b :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)

# 3. Logical Operators
x = True
y = False

print("\nLogical Operators")
print("x and y :", x and y)
print("x or y :", x or y)
print("not x :", not x)
print("not y :", not y)

# 4. Placement Eligibility using Logical Operators
cgpa = 8.2
attendance = 80

print("\nPlacement Eligibility")

# Eligible if CGPA >= 7.5 and Attendance >= 75%
if cgpa >= 7.5 and attendance >= 75:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")
```

```python
model_name = "Logistic Regression"
accuracy = 0.87345

# Old, clunky way (avoid this):
print("Accuracy of " + model_name + " is " + str(accuracy))

# f-string way (use this):
print(f"Accuracy of {model_name} is {accuracy}")

# ':.3f' inside the braces rounds a float to 3 decimal places -
# you will use this in almost every experiment to print clean metrics.
print(f"Accuracy of {model_name} is {accuracy:.3f}")
```

```python
# Program 4: Lists and Indexing

# 1. Create a list
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

# Access elements using indexing
print("List:", fruits)
print("First Element:", fruits[0])
print("Second Element:", fruits[1])
print("Last Element:", fruits[-1])

# 2. Display first, second, and last elements
print("\nDisplaying Elements:")
print("First:", fruits[0])
print("Second:", fruits[1])
print("Last:", fruits[-1])

# 3. List Slicing
print("\nList Slicing:")
print("First Three Elements:", fruits[0:3])
print("Last Two Elements:", fruits[-2:])
print("Elements from Index 1 to 3:", fruits[1:4])

# 4. Add a new element using append()
fruits.append("Pineapple")
print("\nAfter Appending a New Element:")
print(fruits)

# 5. Display the total number of elements
print("\nTotal Number of Elements:", len(fruits))
```

```python
# Program 5: Dictionaries

# 1. Create a dictionary
student = {
    "Name": "John",
    "Age": 20,
    "Course": "B.Tech AI & ML",
    "CGPA": 8.7
}

# Display all key-value pairs
print("Student Dictionary:")
print(student)

# 2. Access the value of a specific key
print("\nAccessing Specific Values:")
print("Name:", student["Name"])
print("CGPA:", student["CGPA"])

# 3. Iterate through the dictionary using items()
print("\nIterating Through Dictionary:")
for key, value in student.items():
    print(key, ":", value)

# 4. Store Machine Learning models in a dictionary
ml_models = {
    "Linear Regression": "Regression Model",
    "Decision Tree": "Tree-Based Model",
    "Random Forest": "Ensemble Model",
    "Support Vector Machine": "Classification Model",
    "K-Nearest Neighbors": "Instance-Based Model"
}

print("\nMachine Learning Models:")
for model, model_type in ml_models.items():
    print(model, "->", model_type)
```

```python
# Program 6: Conditional Statements

# 1. Classify Student CGPA
cgpa = 8.2

print("CGPA Classification")
if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")

# 2. Check for Missing Values in a Dataset
dataset = [10, 20, None, 40, 50]

print("\nChecking for Missing Values")
if None in dataset:
    print("Missing values are present in the dataset.")
else:
    print("No missing values in the dataset.")

# 3. Placement Eligibility
cgpa = 7.8

print("\nPlacement Eligibility")
if cgpa >= 7.5:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")

# 4. Assign Grades Based on Marks
marks = 86

print("\nGrade Assignment")
if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
# Program 7: Loops

# List of numbers
numbers = [12, 45, 67, 23, 89, 34, 56, 78]

# 1. Display all elements using a for loop
print("Elements in the List:")
for num in numbers:
    print(num)

# 2. Calculate the average of numbers in the list
sum = 0
for num in numbers:
    sum += num

average = sum / len(numbers)
print("\nAverage:", average)

# 3. Display values greater than a given threshold
threshold = 50

print("\nValues Greater Than", threshold, ":")
for num in numbers:
    if num > threshold:
        print(num)

# 4. Find the maximum and minimum values using loops
maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print("\nMaximum Value:", maximum)
print("Minimum Value:", minimum)

# 5. Count elements satisfying a condition
count = 0

for num in numbers:
    if num % 2 == 0:      # Condition: Even numbers
        count += 1

print("\nNumber of Even Elements:", count)
```
